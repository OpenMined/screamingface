# Overview

**screamingface lets you combine several language models into one "fusion,"
run it on a benchmark, and measure whether the group beats the best single
model.** It's a research tool for studying *when — and why — many models are
better than one*.

This first notebook is the whole story on a single page: connect, discover,
compose, run, compare. The notebooks that follow (`01`–`05`) each take one step
and go slower.

> **Good to know.** Benchmark *questions* are real (pulled from the HuggingFace
> Hub). Model *answers* are **simulated** deterministically, so everything runs
> offline, for free, and gives the same result every time. Connecting a real
> provider is wired up but doesn't spend money yet.

## The pipeline

```
  ┌───────────┐   ┌───────────┐   ┌──────────────┐   ┌─────────────┐   ┌──────────────┐
  │  CONNECT  │──►│ DISCOVER  │──►│   COMPOSE    │──►│     RUN     │──►│   COMPARE    │
  │ providers │   │  models   │   │   a fusion   │   │ a benchmark │   │  &  share    │
  └───────────┘   └───────────┘   └──────────────┘   └─────────────┘   └──────────────┘
   sf.setup()    sf.models.list()  Fusion()  fusion.evaluate()   sf.leaderboard()
```

Each box is one verb. You can do every step from code (as below) or from a small
interactive widget — they're two doors into the same thing.

### Step 0 · Set up

````python
import warnings; warnings.simplefilter("ignore")
import screamingface as sf
sf.mock_widgets(True)   # render static widget previews (great on GitHub); delete this line to go interactive
sf.__version__
````

### Step 1 · Connect a provider
Providers are where models come from (Anthropic, OpenAI, …). `sf.setup()` opens a
panel to plug them in; you can also connect from code. *(Skipping this still
works here, because answers are simulated.)* → deep dive: **`01_authentication`**.

````python
# sf.connect("anthropic")          # reads your ANTHROPIC_API_KEY
sf.connections()                    # a live view of what's connected
````

### Step 2 · Discover models
Models are named `provider/model`. `sf.models.list(...)` searches and sorts the
catalog and hands back plain ids you can drop straight into a fusion.
→ deep dive: **`02_models`**.

````python
ids = sf.models.list(max_price=20, sort="price"); ids[:5]
````

### Step 3 · Compose a fusion
A fusion runs a **loop** (ask every model) and a **reduce** (combine the
answers into one), with an optional **judge** to break ties.
→ deep dive: **`03_fusions`**.

````python
fusion = sf.Fusion("my-first-fusion",
    models=["anthropic/claude-opus-4.8", "google/gemini-2.5-pro", "openai/gpt-5"],
    reduce="majority_vote")
fusion
````

### Step 4 · Check the cost, then run

````python
fusion.estimate_cost("gpqa", first=60)     # know the spend before you spend it
````

````python
run = fusion.evaluate("gpqa", first=60, seed=0)
run.score, run.gain      # accuracy, and how much it beat the best single model
````

`gain` is the number to watch: **how many points the fusion scored above its
best individual member.** Positive gain means the combination actually helped.

### Step 5 · Compare and share

````python
run.leaderboard()      # where your run lands against baselines & others
````

````python
run.fusion.url       # the whole recipe as one shareable string
````
